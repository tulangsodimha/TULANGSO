```python
from IPython.display import display, HTML

html_content = """
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8" />
    <meta name="viewport" content="width=device-width, initial-scale=1" />
    <title>Biryani Order</title>
    <style>
        body {
            font-family: Arial, sans-serif;
            max-width: 400px;
            margin: 40px auto;
            padding: 20px;
            border: 2px solid #f2a365;
            border-radius: 10px;
            background-color: #fff7f0;
        }
        h1 {
            color: #d35400;
            text-align: center;
        }
        label, select, input {
            display: block;
            margin: 15px 0 10px;
            width: 100%;
            font-size: 1rem;
        }
        button {
            background-color: #d35400;
            color: white;
            padding: 10px;
            border: none;
            cursor: pointer;
            font-weight: bold;
            border-radius: 5px;
            width: 100%;
        }
        button:hover {
            background-color: #e67e22;
        }
        #result {
            margin-top: 20px;
            font-weight: bold;
            font-size: 1.1rem;
            color: #2c3e50;
        }
    </style>
</head>
<body>

    <h1>Biryani Order</h1>

    <label for="biryani">Select Biryani:</label>
    <select id="biryani">
        <option value="150">Chicken Biryani - ₹150</option>
        <option value="200">Mutton Biryani - ₹200</option>
        <option value="100">Veg Biryani - ₹100</option>
        <option value="120">Egg Biryani - ₹120</option>
    </select>

    <label for="quantity">Quantity:</label>
    <input type="number" id="quantity" value="1" min="1" />

    <button onclick="placeOrder()">Place Order</button>

    <div id="result"></div>

    <script>
        function placeOrder() {
            const biryaniSelect = document.getElementById('biryani');
            const price = Number(biryaniSelect.value);
            const biryaniName = biryaniSelect.options[biryaniSelect.selectedIndex].text.split(' - ')[0];
            const quantity = Number(document.getElementById('quantity').value);

            if(quantity < 1 || isNaN(quantity)) {
                alert('Please enter a valid quantity.');
                return;
            }

            const total = price * quantity;

            document.getElementById('result').textContent =
                `You ordered ${quantity} x ${biryaniName}. Total price: ₹${total}`;
        }
    </script>

</body>
</html>
"""

display(HTML(html_content))
```
