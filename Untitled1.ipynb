```python
# Simple Biryani Ordering System

# Menu with prices
menu = {
    "Chicken Biryani": 150,
    "Mutton Biryani": 200,
    "Veg Biryani": 100,
    "Egg Biryani": 120
}

print("Welcome to Biryani Order System!")
print("Menu:")
for item, price in menu.items():
    print(f"{item}: ₹{price}")

# Take user input
order = input("Enter the type of biryani you want to order: ")
quantity = int(input("Enter the quantity: "))

# Calculate total price
if order in menu:
    total = menu[order] * quantity
    print(f"Your order: {quantity} x {order}")
    print(f"Total price: ₹{total}")
else:
    print("Sorry, we don't have that biryani.")
```
